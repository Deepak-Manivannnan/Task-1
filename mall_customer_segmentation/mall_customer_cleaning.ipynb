# Mall Customer Segmentation - Data Cleaning and Preprocessing
## Introduction
This notebook focuses on cleaning and validating the Mall Customer Segmentation dataset. The objective is to review the dataset structure, identify potential data quality issues, standardize column names, and verify the consistency of customer demographic and spending information. The cleaned dataset will be prepared for future analysis and customer segmentation tasks.

### 1. Import Library

```python
import pandas as pd 
```

### 2. Load Dataset

```python
df = pd.read_csv("mall_customer_raw.csv")
```

### 3. Column Name Standardization

```python
df.columns = (df.columns.str.lower().str.replace(" ", "_").str.replace("(", "", regex=False).str.replace(")", "", regex=False))
```
