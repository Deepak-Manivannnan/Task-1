# Sales Data - Data Cleaning and Preprocessing
## Introduction
This notebook focuses on cleaning and preprocessing the Sales Export 2019–2020 dataset. The objective is to review dataset structure, standardize column names, correct data types, and validate overall data quality. The cleaned dataset will be prepared for future sales analysis and reporting.

### 1. Import Library

```python
import pandas as pd
```

### 2. Load Dataset

```python
df = pd.read_csv("sales_data_raw.csv")
```

### 3. Dataset Overview

```python
df.head()
```

```python
df.describe(include="all")
```

```python
df.info()
```

### 4. Column Name Standardization

```python
df.columns = (df.columns.str.strip().str.lower())
```

### 5. Data Type Validation and Correction

```python
df["order_value_eur"] = (df["order_value_eur"].str.replace(",", "", regex=False).astype(float))
```

```python
df["date"] = pd.to_datetime(df["date"])
```
