# Customer Personality Analysis – Data Cleaning and Preprocessing
## Introduction
This notebook focuses on cleaning and preprocessing the Customer Personality Analysis dataset. The objective is to identify and resolve common data quality issues such as incorrect data types, missing values, invalid categorical entries, and duplicate records. The cleaned dataset produced from this process will be more reliable and suitable for further analysis and visualization.

### 1. Import Libraries

```python
import pandas as pd
```

### 2. Load Dataset

```python
df = pd.read_csv("customer_personality_raw.csv", sep="\t")
```

### 3. Dataset Overview

```python
df.head()
```

```python
df.describe(include='all')
```

```python
df.info()
```

### 4. Column Name Standardization

```python
df.columns = df.columns.str.lower().str.strip()
```

### 5. Data Type Correction

```python
df["dt_customer"] = pd.to_datetime(df["dt_customer"], dayfirst=True)
```

### 6. Handling Null Values

```python
df.isnull().sum()
```

```python
(df["income"].isnull().sum() / len(df)) * 100
```

```python
df = df.dropna(subset=["income"])
```

### 7. Categorical Data Inspection and Standardization

```python
df["education"].unique()
```

```python
df["marital_status"].unique()
```

```python
df["marital_status"].value_counts()
```

```python
df = df[(df["marital_status"] != "Absurd") & (df["marital_status"] != "YOLO")]
```

### 8. Duplicate Record Check

```python
df.duplicated().sum()
```

### 9. Final Data Quality Validation

```python
df.info()
```

### 10. Export Cleaned Dataset

```python
df.to_csv("customer_personality_cleaned.csv", index=False)
```
