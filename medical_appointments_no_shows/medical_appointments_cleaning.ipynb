# Medical Appointment No-Shows - Data Cleaning and Preprocessing
## Introduction
This notebook focuses on cleaning and preprocessing the Medical Appointment No-Shows dataset. The objective is to identify and resolve data quality issues such as invalid values, inconsistent date formats, duplicate records, inconsistent values. The cleaned dataset produced from this process will be more reliable and suitable for further analysis and reporting.


### 1. Import Libraries

```python
import pandas as pd
```

### 2. Load Dataset

```python
df = pd.read_csv("medical_appointments_raw.csv")
```

### 3. Dataset Overview

```python
df.head()
```

```python
df.describe(include="all")
```

```python
df.info()
```

### 4. Column Name Standardization

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df = df.rename(columns={
    'patientid': 'patient_id',
    'appointmentid': 'appointment_id',
    'scheduledday': 'scheduled_day',
    'appointmentday': 'appointment_day',
    'handcap': 'handicap',
    'no-show': 'no_show'
})
```

### 5. Data Type Validation and Correction

```python
df["scheduled_day"] = pd.to_datetime(df["scheduled_day"])

df["appointment_day"] = pd.to_datetime(df["appointment_day"])
```

### 6. Data Quality Check and Correction

```python
df[df["scheduled_day"].dt.date > df["appointment_day"].dt.date]
```

```python
invalid_dates = df[df["scheduled_day"].dt.date > df["appointment_day"].dt.date].index

df = df.drop(invalid_dates)
```

```python
df["age"].describe()
```

```python
df[df["age"] < 0]
```

```python
df = df[df["age"] >= 0]
```

### 7. Handling missing Values

```python
df.isnull().sum()
```
